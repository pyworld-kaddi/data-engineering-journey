```python
def greet():
    return 'Hello Saravanan'

print(greet())
```

```python
def add(a,b):
    return a+b
print(add(10,20))
```

```python
def is_pass(mark):
    res="Fail"
    if mark > 50:
        res = "Pass"
    return res

print(is_pass(72))
print(is_pass(30))
```

```python
def square(num):
    return num*num
print(square(5))
```

```python
def bigger(a,b):
    big=b
    if a>b:
        big=a
    return big
print(bigger(10,20))
print(bigger(50,30))
```

```python
def count_passed(marks):
    counts=0
    for i in marks:
        if int(i) > 50:
            counts+=1
    return counts

print(count_passed([45, 67, 90, 23, 88]))
```

```python
def total_marks(marks):
    total=0
    for i in marks:
        total+=i
    return total

print(total_marks([45, 67, 90, 23, 88]))
```

```python
def average_marks(marks):
    total=0
    counts=0
    for i in marks:
        total+=i
        counts+=1
    return total/counts

print(average_marks([45, 67, 90, 23, 88]))
```

```python
def highest_marks(marks):
    high=marks[0]
    for i in marks:
        if i > high:
            high = i
    return high

print(highest_marks([45, 67, 90, 23, 88]))
```
