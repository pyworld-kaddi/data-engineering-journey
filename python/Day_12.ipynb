```python
class Student:

    def __init__(self,name,mark):
        self.name = name
        self.mark = mark
    
    def grade(self):
        if self.mark >= 90:
            return 'A'
        elif self.mark >= 75:
            return 'B'
        else:
            return 'C'

    def is_pass(self):
        if self.mark >= 50:
            return 'PASS'
        else:
            return "FAIL"
    
    def student_report(self):
        print('NAME:',self.name)
        print('MARK:',self.mark)
        print('GRADE:',self.grade())
        print('STATUS:',self.is_pass())

student1=Student("Ram", 85)
student2=Student("Arun",40)
student3=Student("John",95)

student=[student1,student2,student3]

for a in student:
    a.student_report()
    print()
```
