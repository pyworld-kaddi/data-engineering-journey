```python
def high_salary_employees(employees):
    result=[]
    for i in employees:
        if i["salary"]>=25000:
            result.append(i["name"])
    return result

print(high_salary_employees([
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000},
    {"name": "John", "salary": 15000}
]))
    
```

```python
def total_salary(employees):
    result=0
    for i in employees:
        result+=i["salary"]
    return result

print(total_salary([
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000},
]))
    
```

```python
def average_salary(employees):
    result=0
    counts=0
    for i in employees:
        result+=i["salary"]
        counts+=1
    return result/counts

print(average_salary([
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000},
]))
    
```

```python
def highest_salary_employee(employees):
    result=employees[0]["name"]
    high=employees[0]["salary"]
    for i in employees:
        if i["salary"]> high:
            result = i["name"]
            high = i["salary"]
    return result

print(highest_salary_employee([
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000},
    {"name": "John", "salary": 15000}
]))
```

```python
def count_high_salary(employees):
    result=0
    
    for i in employees:
        if i["salary"]>=25000:
            result += 1
    return result

print(count_high_salary([
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000},
    {"name": "John", "salary": 15000}
]))
```

```python
def employee_summary(employees):
    
    result={}
    ts=0
    ec=0
    hse=[]
    hsec=0
    htse=employees[0]["name"]
    high=employees[0]["salary"]
    
    for i in employees:
        ts+=i["salary"]
        ec+=1
        if i["salary"]>=25000:
            hse.append(i["name"])
            hsec+=1
        if i["salary"]> high:
            htse = i["name"]
            high = i["salary"]
    
    result["total_salary"]=ts
    result["employee_count"]=ec
    result["average_salary"]=ts/ec
    result["high_salary_emplyoees"]=hse
    result["count_high_salary_employees"]=hsec
    result["highest_salary_employee"]=htse
    
    return result

print(employee_summary([
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000},
    
]))
```
