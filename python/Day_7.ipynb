```python
def student_report(students):
    
    ts=0;ps=[];fs=[];avg=[];result={}
   
    for i in students:
        ts+=1
        ca=0
        tca=0
        passed=True
        
        for m in i["marks"]:
            ca+=m
            tca+=1
            
            if m < 50:
                passed=False
            
        
        if passed:
            ps.append(i["name"])
        else:
            fs.append(i["name"])
        
        avg.append({"name":i["name"],"avg_mar":ca/tca})
    
    condition=len(avg)-1
    hasm=avg[0]["avg_mar"]
    has=avg[0]["name"]
    caa=0
    
    while condition >=0:
        
        if avg[condition]["avg_mar"]>hasm:
            hasm=avg[condition]["avg_mar"]
            has=avg[condition]["name"]
        
        caa+=avg[condition]["avg_mar"]
        condition-=1
        
    result["total_students"]=ts
    result["passed_students"]=ps
    result["failed_students"]=fs
    result["highest_average_student"]=has
    result["class_average"]=caa/ts
    
    return result
    
        

print(student_report([
    {"name": "Arun", "marks": [85, 90, 78]},
    {"name": "Bala", "marks": [45, 50, 40]},
    {"name": "Kumar", "marks": [72, 68, 80]},
    {"name": "John", "marks": [30, 25, 40]}
]))
```
