```python
class Engine:

    def start(self):
        print("Engine Started")

class Car:

    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car Started")

car1 = Car()
car1.start_car()
```

```python
class Battery:

    def charge(self):
        print("Battery Charging")

class Mobile:

    def __init__(self):
        self.battery = Battery()

    def charge_mobile(self):
        self.battery.charge()
        print("Mobile Ready")

mobile1 = Mobile()

mobile1.charge_mobile()
```

```python
class Book:
    def __init__(self,title):
        self.title=title

class Library:
    def __init__(self,name,book):
        self.name = name
        self.book = Book(book)

    def search(self):
        print (self.name,"contains",self.book.title)

book1 = Library("City Library","Python Basics")

book1.search()
        
```

```python
class Teacher():
    def __init__(self,name):
        self.teacher_name = name
class School():
    def __init__(self,name,teacher):
        self.school_name = name
        self.teacher = Teacher(teacher)

school1 = School("ABC School","Arun")

print(school1.school_name,"employes",school1.teacher.teacher_name)
```

```python
class Student():
    def __init__(self,name):
        self.student_name = name
class College():
    def __init__(self,name,student):
        self.college_name = name
        self.student = student
    def details(self):
        print(self.student.student_name,"studies in",self.college_name)
        
student1=Student("Arun")
college1=College("ABC College",student1)

college1.details()
```

```python
class Department():
    def __init__(self,name):
        self.department_name = name
class Company():
    def __init__(self,name,department):
        self.company_name = name
        self.department = Department(department)
    def details(self):
        print(self.company_name,"has a dependent",self.department.department_name)
        

company1=Company("ABC Company","Stores department")

company1.details()
```

```python
class Doctor():
    def __init__(self,name):
        self.doctor_name = name
class Hospital():
    def __init__(self,name,doctor):
        self.hospital_name = name
        self.doctor = doctor
    def details(self):
        print(self.hospital_name,"has",self.doctor.doctor_name)
        
doctor1=Doctor("Doctor Arun")
hospital1=Hospital("City Hospital",doctor1)

hospital1.details()
```

```python
class Driver():
    def __init__(self,name):
        self.driver_name = name
class Engine():
    def __init__(self,name):
        self.engine_name = name
class Vehicle():
    def __init__(self,name,engine,driver):
        self.vehicle_name = name
        self.engine = Engine(engine)
        self.driver = driver
    def driver_details(self):
        print(self.vehicle_name,"is driven by",self.driver.driver_name)
    def engine_details(self):
        print(self.vehicle_name,"uses",self.engine.engine_name)
        
driver1=Driver("Arun")
vehicle1=Vehicle("Car","Petrol Engine",driver1)

vehicle1.engine_details()
vehicle1.driver_details()
```

```python
class Remote:
    def __init__(self,func):
        self.func = func
    def function(self):
        if self.func == "on":
            return self.power_on()
        else:
            return self.power_off()
    def power_on(self):
        check_tv_circuit=True
        signal_Tranfission = True
        internal_processing = "on process"
        return "TV Turned ON"
    def power_off(self):
        check_tv_circuit=False       
        signal_Tranfission = False
        internal_proceessing = "off process"
        return "TV Turned OFF"

func1=Remote("on")
func2=Remote("off")

print(func1.function())
print(func2.function())
    
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):

    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
       print("Meow")

dog1 = Dog()
cat1 = Cat()

dog1.sound()
cat1.sound()
```
