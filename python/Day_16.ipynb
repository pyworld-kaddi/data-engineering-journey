```python
class Engine:

    def start(self):
        print("Engine Started")

class Car:

    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car Started")

car1 = Car()
car1.start_car()
```

```python
class Battery:

    def charge(self):
        print("Battery Charging")

class Mobile:

    def __init__(self):
        self.battery = Battery()

    def charge_mobile(self):
        self.battery.charge()
        print("Mobile Ready")

mobile1 = Mobile()

mobile1.charge_mobile()
```

```python
class Laptop:
    
    def __init__(self,b_name):
        self.b_name = b_name

    def brand(self):
        return self.b_name
   
class Employee:

    def __init__(self,name,bname):
        self.name = name
        self.laptop=Laptop(bname)
        
        

    def details(self):
        print(self.name,"uses",self.laptop.brand())


employee1 = Employee("Arun","Dell")



employee1.details()

        
```

```python
class Company:
    
    def __init__(self,e_name):
        self.e_name = e_name

    def engine(self):
        return self.e_name
   
class Car:

    def __init__(self,name,ename):
        self.name = name
        self.company = Company(ename)
        
        

    def details(self):
        print(self.name,"uses",self.company.engine())


car1 = Car("BMW","Bosch Engine")



car1.details()

        
```

```python
class Book:
    def __init__(self,title):
        self.title=title

class Library:
    def __init__(self,name,book):
        self.name = name
        self.book = Book(book)

    def search(self):
        print (self.name,"contains",self.book.title)

book1 = Library("City Library","Python Basics")

book1.search()
        
```

```python
class Teacher():
    def __init__(self,name):
        self.teacher_name = name
class School():
    def __init__(self,name,teacher):
        self.school_name = name
        self.teacher = Teacher(teacher)

school1 = School("ABC School","Arun")

print(school1.school_name,"employes",school1.teacher.teacher_name)
```
