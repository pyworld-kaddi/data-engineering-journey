```python
text = "Data Engineering with Python"

vowel_count=0;first_word="";words={};word_count=1;space_count=0;condition=0

while condition < len(text):
    
    if text[condition] in "aeiouAEIOU":
        vowel_count+=1
    
    if text[condition] == " ":
        words[word_count]=first_word
        first_word=""
        word_count+=1
        space_count+=1
        condition+=1
        continue

    first_word+=text[condition]
    condition+=1

words[word_count]=first_word
longest_word_len=0
for i in words:
    if len(words[i]) > longest_word_len:
        longest_word_len=len(words[i])
        longest_word=words[i]

result={
    "word_count": word_count,
    "char_count": len(text),
    "space_count": space_count,
    "vowel_count": vowel_count,
    "longest_word": longest_word
}

print(result)
```

```python
text = "Data Engineering with Python"

word="";words=[];word_count=1;condition=0

while condition < len(text):
    
    if text[condition] == " ":
        words.append(word)
        word_count+=1
        word=""
        condition+=1
        continue
    
    word+=text[condition]
    condition+=1
    
words.append(word)   
longest_word_len=0
shortest_word_len=len(text)
for i in words:
    if len(i) > longest_word_len:
        longest_word_len=len(i)
        longest_word=i
    if len(i) <= shortest_word_len:
        shortest_word_len=len(i)
        shortest_word=i

result={
    "word_count": word_count,
    "words": words,
    "longest_word": longest_word,
    "shortest_word": shortest_word
}

print(result)
```
