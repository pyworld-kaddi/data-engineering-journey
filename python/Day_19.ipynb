```python
class Book:
    
    def __init__(self,title,author,available):
        self.title = title
        self.author = author
        self.available = available
        
    def __str__(self):
        return f"{self.title} - {self.status()}"

    def status(self):
        if self.available:
            return 'Available'
        else:
            return 'Not Available' 
    def borrow(self):
        if self.status() == "Available":
            self.available = False
            return "Book Borrowed"
        else:
            return "Book Already Borrowed"
    

book1 = Book("Python Basics", "Arun", True)
book2 = Book("SQL Guide", "Bala", False)

books=[book1,book2]

for b in books:
    print(b.borrow())
```

```python
class Book:
    
    def __init__(self,title,author,available):
        self.title = title
        self.author = author
        self.available = available
        
    def __str__(self):
        return f"{self.title} - {self.status()}"

    def status(self):
        if self.available:
            return 'Available'
        else:
            return 'Not Available' 
    def borrow(self):
        if self.status() == "Available":
            self.available = False
            return "Book Borrowed"
        else:
            return "Book Already Borrowed"

    def return_book(self):
        if self.available == False:
            self.available = True
            return "Book Returned"
        else:
            return "Book Already Available"
    

book1 = Book("Python Basics", "Arun", True)
book2 = Book("SQL Guide", "Bala", False)

books=[book1,book2]

print(book1.borrow())
print(book1.borrow())
print(book1.return_book())
print(book1.return_book())
```
