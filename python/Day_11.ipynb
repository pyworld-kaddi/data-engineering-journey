```python
class student:

    def __init__(self,name,age,mark):
        self.name = name
        self.age = age
        self.mark = mark
    
    def show(self):
        print(self.name)
        print(self.age)
        print(self.mark)
        print()

    def is_pass(self):
        if self.mark >= 50:
            print(self.name,"PASS")
        else:
            print(self.name,"Fail")

student1=student("Ram",20,85)
student2=student("Arun",22,40)

student1.is_pass()
student2.is_pass()
```

```python
class Student:

    def __init__(self,name,age,mark):
        self.name = name
        self.age = age
        self.mark = mark
    
    def show(self):
        print(self.name)
        print(self.age)
        print(self.mark)
        print()

    def is_pass(self):
        if self.mark >= 50:
            return "PASS"
        else:
           return "Fail"

student1=Student("Ram",20,85)
student2=Student("Arun",22,40)

print(student1.name, "-", student1.is_pass())
print(student2.name, "-", student2.is_pass())
```

```python
class Employee:

    def __init__(self,name,age,salary):
        self.name = name
        self.age = age
        self.salary= salary
    
    def show(self):
        print(self.name)
        print(self.age)
        print(self.salary)

    def is_bonus(self):
        if self.salary >= 60000:
            return 10000
        else:
           return 5000

    def salary_category(self):
        if self.salary >= 70000:
            return "High"
        elif self.salary >= 60000:
            return "Medium"
        else:
            return "Low"

    def employee_report(self):
        print("Name:",self.name)
        print("Age:",self.age)
        print("Salary:",self.salary)
        print("Category:",self.salary_category())
        print("Bonus:",self.is_bonus())

    def annual_salary(self):
        return self.salary * 12

employee1=Employee("Ram",20,50000)
employee2=Employee("Arun",22,65000)
employee3=Employee("john",21,70000)
l=[employee1,employee2,employee3]

for a in l:
    print(a.name,"-",a.annual_salary())
    
```
