```python
read_file=open("students.txt","r")
write_file=open("passed_students.txt","a")

for line in read_file:
    parts=line.strip().split(",")
    if int(parts[1])>50:
        write_file.write(parts[0]+"\n")
        

write_file.close()
read_file.close()
```
