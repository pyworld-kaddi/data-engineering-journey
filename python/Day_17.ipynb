```python
student = {
    "name": "Arun",
    "mark": 85
}

import json

json_data = json.dumps(student)

print(json_data)
print(type(json_data))
```

```python
employee = {
    "name": "Arun",
    "salary": 50000
}

import json

json_data1 = json.dumps(employee)
json_data2 = json.loads(json_data1)

print(json_data1)
print(type(json_data1))

print(json_data2)
print(type(json_data2))
```

```python
employee = {
    "name": "Arun",
    "salary": 50000,
    "city":"chennai"
}

import json

json_data1 = json.dumps(employee)
json_data2 = json.loads(json_data1)

print("Name:", json_data2["name"])
print("Salary:", json_data2["salary"])
print("City:", json_data2["city"])
```
