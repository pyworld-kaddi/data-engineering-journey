```python
employee = {
    "name": "Arun",
    "salary": 50000,
    "city":"chennai"
}

import json
file = open("employee.json",'w')
json.dump(employee,file)
file=open("employee.json",'r')
data=json.load(file)

print(data["name"])
print(data["salary"])
print(data["city"])

```

```python
employee = [{"name": "Arun","salary": 50000},
            {"name": "Ram","salary": 60000},
            {"name": "John","salary": 70000}]
           

import json
with open("employee.json", "w") as file:
    json.dump(employee,file)
with open("employee.json", "r") as file:
    data=json.load(file)

for l in data:
    print(l["name"])

```

```python
employee = [{"name": "Arun","salary": 50000},
            {"name": "Ram","salary": 60000},
            {"name": "John","salary": 70000}]
           

import json
with open("employee.json", "w") as file:
    json.dump(employee,file)
with open("employee.json", "r") as file:
    data=json.load(file)

for emp in data:
    if emp["salary"] >= 60000:
        print(emp["name"])

```

```python
employee = [{"name": "Arun","salary": 50000},
            {"name": "Ram","salary": 60000},
            {"name": "John","salary": 70000}]
           

import json
with open("employee.json", "w") as file:
    json.dump(employee,file)
with open("employee.json", "r") as file:
    data=json.load(file)
total=0
for emp in data:
    total += emp["salary"]
print("Total Salary:",total)

```

```python
employee = [{"name": "Arun","salary": 50000},
            {"name": "Ram","salary": 60000},
            {"name": "John","salary": 70000}]
           

import json
with open("employee.json", "w") as file:
    json.dump(employee,file)
with open("employee.json", "r") as file:
    data=json.load(file)

high=0
name=""
for emp in data:
    if emp["salary"] > high:
        high = emp["salary"]
        name = emp["name"]

print("Highest Salary Employee:",name)
print("Salary:" ,high)

```
