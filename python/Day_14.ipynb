```python
numbers = [1, 2, 3, 4, 5]

result = [x * 2 for x in numbers]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

result = [x for x in numbers if x % 2 == 0]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

result = [x*x for x in numbers]

print(result)
```

```python
names = ["arun", "bala", "kumar"]

result = [name.upper() for name in names]

print(result)
```

```python
numbers = [10, 15, 20, 25, 30]

result = [x for x in numbers if x > 20]

print(result)
```

```python
square = lambda x: x * x

print(square(5))
```

```python
add = lambda a, b: a + b

print(add(10, 20))
print(add(5, 7))
```

```python
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
```

```python
numbers = [50, 20, 40, 10, 30]

print(sorted(numbers))
```

```python
numbers = [50, 20, 40, 10, 30]

print(sorted(numbers, reverse=True))
```

```python
names = ["Kumar", "Arun", "Bala"]

print(sorted(names))
```

```python
employees = [
    ("Arun", 50000),
    ("Bala", 65000),
    ("Kumar", 55000)
]

print(sorted(employees, key=lambda x: x[1]))
```

```python
students = [
    ("Arun", 85),
    ("Bala", 92),
    ("Kumar", 78),
    ("John", 95)
]

result = sorted(
    students,
    key=lambda x: x[1],
    reverse=True
)

print(result[0])
```

```python
employees = [
    ("Arun", 50000),
    ("Bala", 65000),
    ("Kumar", 55000),
    ("John", 70000)
]

paid = sorted(
    employees,
    key=lambda x: x[1],
    reverse=True)
salary=0
for x in employees:
    salary+=x[1]

result={
    "highest_paid":paid[0] ,
    "lowest_paid": paid[-1],
    "average_salary": salary/(len(employees))
}

print(result)
```

```python
employees = [
    ("Arun", 50000),
    ("Bala", 65000),
    ("Kumar", 55000),
    ("John", 70000)
]

result = sorted(
    employees,
    key=lambda x: x[0]
)

print(result)
```

```python
employees = [
    ("Arun", 50000),
    ("Bala", 65000),
    ("Kumar", 55000),
    ("John", 70000)
]

result = sorted(
    employees,
    key=lambda x: len(x[0])
)

print(result)
```
