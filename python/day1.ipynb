```python
employees = [
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000}
]
for i in employees:
    if i['salary']<20000:
        print(i['name'],' gets bonus')
    else:
        print(i["name"]," no bonus")
```

```python
employees = [
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 18000},
    {"name": "Kumar", "salary": 32000}
]
bonus_employees=[]
for i in employees:
    if i['salary']<20000:
        bonus_employees.append(i['name'])
print(bonus_employees)
```
