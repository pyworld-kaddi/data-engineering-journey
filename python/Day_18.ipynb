```python
from datetime import datetime

join_date = "17-07-2024"
today = "20-07-2026"
date_object_1 = datetime.strptime(join_date, "%d-%m-%Y")
date_object_2 = datetime.strptime(today, "%d-%m-%Y")
print (date_object_2 - date_object_1)
```
