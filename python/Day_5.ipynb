```python
def odd_numbers(numbers):
    result=[]
    for i in numbers:
        if i%2!=0:
            result.append(i)
    return result
print(odd_numbers([1,2,3,4,5,6,7,8]))
```

```python
def numbers_greater_than(numbers, limit):
    result=[]
    for i in numbers:
        if i > limit:
            result.append(i)
    return result
print(numbers_greater_than([10,20,30,40,50], 25))
```

```python
def count_greater_than(numbers, limit):
    result=0
    for i in numbers:
        if i > limit:
            result+=1
    return result
print(count_greater_than([10,20,30,40,50], 25))
```

```python
def square_numbers(numbers):
    result=[]
    for i in numbers:
        result.append(i*i)
    return result
print(square_numbers([1,2,3,4,5]))
```

```python
def total_even_numbers(numbers):
    result=0
    for i in numbers:
        if i%2==0:
            result+=i
    return result
print(total_even_numbers([1,2,3,4,5,6]))
```

```python
def passed_students(marks):
    result=[]
    for i in marks:
        if i > 50:
            result.append(i)
    return result
print(passed_students([45,67,90,23,88]))
```

```python
def failed_students(marks):
    result=[]
    for i in marks:
        if i < 50:
            result.append(i)
    return result
print(failed_students([45,67,90,23,88]))
```

```python
def passed_students_names(students):
    result=[]
    for i in students:
        if int(i[1])> 50:
            result.append(i[0])
    return result
print(passed_students_names([("Arun", 85),("Bala", 45),("Kumar", 72),("John", 30)]))
```

```python
def average_student_marks(students):
    counts=0
    result=0
    for i in students:
        result +=i[1]
        counts+=1
    return result/counts
print(average_student_marks([("Arun", 85),("Bala", 45),("Kumar", 72),("John", 30)]))
```
