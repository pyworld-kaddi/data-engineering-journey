```python
employees = ["Arun","Bala","Arun","John","Bala","Kumar"]
unique_employees=set(employees)
result={
    "total_records": len(employees),
    "unique_employees": unique_employees,
    "unique_count": len(unique_employees)
}
print(result)
```

```python
sales_team = {"Arun", "Bala", "John", "Kumar"}
it_team = {"John", "David", "Kumar", "Raj"}
 
result={
    "all_employees": sales_team | it_team,
    "common_employees": sales_team & it_team,
    "sales_only": sales_team - it_team,
    "it_only": it_team - sales_team  
}
print(result)
```
