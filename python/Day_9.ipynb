```python
file = open("marks.csv", "r")
total_marks=0
total_lines=0
highest_student=""
highest_mark=0

for line in file:
    total_lines +=1
    parts = line.strip().split(",")
    total_marks += int(parts[1])

    if highest_mark< int(parts[1]):
        highest_student=parts[0]
        highest_mark=int(parts[1])

result={"Total Marks":total_marks,
        "Average Marks":total_marks/total_lines,
        "Highest Student":highest_student,
        "Highest Mark":highest_mark
}

print(result)
file.close()
```

```python
file = open("employees.csv", "r")

total_salary=0
total_employees=0
highest_paid_employee=""
highest_salary=0

for line in file:
    total_employees +=1
    parts = line.strip().split(",")
    salary = int(parts[1])
    total_salary += salary

    if highest_salary < salary:
        highest_paid_employee=parts[0]
        highest_salary= salary

result={"Total Salary":total_salary,
        "Employees Count":total_employees,
        "Average Salary":total_salary/total_employees,
        "Highest Salary":highest_salary,
        "Highest Paid Employee":highest_paid_employee
}

print(result)
file.close()
```

```python
file = open("employees.csv", "r")

total_salary=0
total_employees=0
lines=file.readline()
lines=lines.strip().split(",")
highest_paid_employee=lines[0]
highest_salary=int(lines[1])
lowest_paid_employee=lines[0]
lowest_salary=int(lines[1])



for line in file:
    total_employees +=1
    parts = line.strip().split(",")
    salary = int(parts[1])
    total_salary += salary

    if highest_salary < salary:
        highest_paid_employee=parts[0]
        highest_salary= salary

    if lowest_salary > salary:
        lowest_paid_employee=parts[0]
        lowest_salary= salary

result={"Total Salary":total_salary,
        "Employees Count":total_employees,
        "Average Salary":total_salary/total_employees,
        "Highest Salary":highest_salary,
        "Highest Paid Employee":highest_paid_employee,
        "Lowest Paid Employee": lowest_paid_employee,
        "Lowest Salary": lowest_salary
}

print(result)
file.close()
```

```python
file=open("sales.csv","w")
file.write("Arun,12000\n"
"Bala,18000\n"
"Kumar,15000\n"
"John,22000\n"
"David,9000")
file.close()
```

```python
file = open("sales.csv", "r")

total_sales=0
sales_people_count=0

highest_sales_person=""
highest_sales=0
lowest_sales_person=""
lowest_sales=float('inf')



for line in file:
    sales_people_count +=1
    parts = line.strip().split(",")
    sales = int(parts[1])
    total_sales += sales

    if highest_sales < sales:
        highest_sales_person=parts[0]
        highest_sales= sales

    if lowest_sales > sales:
        lowest_sales_person=parts[0]
        lowest_sales= sales

result={"Total Sales":total_sales,
        "Average Sales":total_sales/sales_people_count,
        "Highest Sales Person":highest_sales_person,
        "Highest Sales":highest_sales,
        "Lowest Sales Person": lowest_sales_person,
        "Lowest Sales": lowest_sales,
        "Sales Person Count":sales_people_count
}

print(result)
file.close()
```
