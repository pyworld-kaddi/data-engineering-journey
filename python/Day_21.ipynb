```python
class Person:

    def show_role(self):
        print("I am a Person")

class Employee(Person):

    def show_role(self):
        print("I am an Employee")

employee1 = Employee()

employee1.show_role()
```

```python
class Person:

    def introduce(self):
        print("I am a Person")

class Employee(Person):

    def introduce(self):
        print("I am an Employee")

class Student(Person):

    def introduce(self):
        print("I am a Student")

student1 = Student()
employee1 = Employee()

student1.introduce()
employee1.introduce()
```

```python
class Person:

    def work(self):
        print("Person is working")

class Student(Person):

    def work(self):
        print("Student is studying")

class Employee(Person):

    def work(self):
        print("Employee is working in office")

student1 = Student()
employee1 = Employee()
person1 = Person()

people = [student1, employee1,person1]

for p in people:
    p.work()
```

```python
class Employee:
    
    def calculate_bonus(self):
        return 5000

class Manager(Employee):
    
    def calculate_bonus(self):
        return 10000

employee1 = Employee()
manager1 = Manager()

workers = [employee1, manager1]

for w in workers:
    print(w.calculate_bonus())
```

```python
class Person:

    def show_name(self):
        print("Arun")

class Worker:

    def work(self):
        print("Working")

class Employee(Person, Worker):
    pass

employee1 = Employee()

employee1.show_name()
employee1.work()
```

```python
class Person:

    def show_name(self):
        print("Arun")

class Worker:

    def work(self):
        print("Working")

class Employee(Person, Worker):

    def salary(self):
        print("Salary = 50000")

employee1 = Employee()

employee1.show_name()
employee1.work()
employee1.salary()
```

```python
class Father:

    def show(self):
        print("Father")

class Mother:

    def show(self):
        print("Mother")

class Child(Father, Mother):
    pass

child1 = Child()

child1.show()
```

```python
class Employee:

    def login(self):
        print("Employee Login")

class Admin:

    def login(self):
        print("Admin Login")

class Manager(Employee, Admin):
    pass

manager1 = Manager()
manager1.login()
```
